```python
from collections import Counter
from matplotlib.colors import to_rgb
from matplotlib.patches import ConnectionPatch

# Colorblind-safe, fixed order: neighboring wedges stay apart under protan/deutan/tritan.
PALETTE = ['#2a78d6', '#eb6834', '#1baf7a', '#eda100', '#e87ba4', '#008300', '#4a3aa7', '#e34948']
GRAY = '#c8c7bd'
INK = 'k'
LEADER = 'k'  # leaders recede; the bold labels carry the eye
STUB = 1.16     # radius a leader breaks clear of the rim at before heading for its label
R_LABEL = 1.28  # labels ride just off the rim, over their own wedge
SMALL = 12      # only for choosing the rotation: which wedges are likely to need a branch


def as_str(col):
    """MaskedColumn -> strings, with masked/placeholder entries as ''."""
    filled = col.filled('') if hasattr(col, 'filled') else col
    vals = np.array([str(v).strip() for v in np.asarray(filled, dtype=str)])
    vals[np.isin(vals, ('--', 'nan', 'None', 'N/A'))] = ''
    return vals


def tex(s):
    """Escape what usetex chokes on (an unescaped % comments out the rest of the label).

    A label that already contains $ is taken to be hand-written LaTeX and left alone --
    that is the only way to get >, <, \\geq into a label, since text mode mangles them.
    """
    if '$' in s:
        return s
    for char in ('\\', '%', '_', '&', '#', '$', '{', '}'):
        s = s.replace(char, rf'\{char}')
    return s


def bold(s):
    """usetex ignores fontweight, so bold has to go through LaTeX, one line at a time."""
    return rf'\textbf{{\boldmath {s}}}' if plt.rcParams['text.usetex'] else s


def on_wedge_ink(color):
    """Dark ink on a light wedge, white on a dark one."""
    r, g, b = to_rgb(color)
    return INK if 0.299 * r + 0.587 * g + 0.114 * b > 0.6 else 'white'


def in_wedge(xy, wedge):
    """Is this point inside the wedge (angles and radii as matplotlib stores them)?"""
    x, y = xy
    r = np.hypot(x, y)
    if r > wedge.r or (wedge.width is not None and r < wedge.r - wedge.width):
        return False
    if wedge.theta2 - wedge.theta1 >= 360 - 1e-9:
        return True  # a lone slice is the whole disc; mod 360 would make its span empty
    ang, lo, hi = np.degrees(np.arctan2(y, x)) % 360, wedge.theta1 % 360, wedge.theta2 % 360
    return lo <= ang <= hi if lo <= hi else (ang >= lo or ang <= hi)


def pie(ax, counts, colors, startangle=None, label_side=1, branch_sep=' -- ', width=None,
        radius=1, stack=None):
    """Pie with labels on the wedges, branching out to the ones too thin to hold one.

    branch_sep joins a branched label to its counts: ' -- ' keeps it on one line, '\\n'
    puts the counts underneath. Labels inside a wedge always stack, to stay narrow.
    width draws a ring of that thickness instead of a full pie and radius scales the whole
    thing, so a second ring can sit outside the first. A blank label is left undrawn, which
    is how a ring covers only part of the circle: give the rest a blank name and no colour.
    Pass the same stack dict to two calls on one axes and their labels queue up together
    instead of both starting at the top and landing on each other.
    """
    total = sum(counts.values())
    vals = list(counts.values())
    if startangle is None:
        # Spin the pie so the run of wedges too thin to hold their own labels sits in the top
        # corner the labels read towards: starting at 12 o'clock for labels running right,
        # ending at 12 o'clock for labels running left. Either way they never have to reach
        # across the pie, whatever the mix of slice sizes.
        small = [i for i, n in enumerate(vals) if 100 * n / total <= SMALL]
        through = (small[0] if label_side > 0 else small[-1] + 1) if small else 0
        startangle = 90 + 360 * sum(vals[:through]) / total
    wedges, _ = ax.pie(vals, colors=colors, startangle=startangle, counterclock=False,
                       radius=radius, wedgeprops=None if width is None else dict(width=width))
    r_inside = 0.55 * radius if width is None else radius - width / 2  # mid-ring, for a ring
    # Pad the side the branch labels run off to, and leave room above for the stack.
    ax.set_xlim(-1.1 * radius, 2.15 * radius) if label_side > 0 else \
        ax.set_xlim(-2.15 * radius, 1.1 * radius)
    ax.set_ylim(-1.3 * radius, 1.4 * radius)

    ax.figure.canvas.draw()  # a renderer, so a label can be measured before it is kept
    renderer = ax.figure.canvas.get_renderer()

    branches = []
    for wedge, color, (label, n) in zip(wedges, colors, counts.items()):
        pct = 100 * n / total
        pct_str = f'{pct:.2f}' if pct < 1 else f'{pct:.1f}'
        tail = f'{n:,} ({pct_str}\\%)'
        theta = np.deg2rad((wedge.theta1 + wedge.theta2) / 2)
        x, y = np.cos(theta), np.sin(theta)
        if not label.strip():  # a spacer arc, drawn but never named
            continue
        # Try the label inside its own slice; branch it out if it would spill into a neighbour.
        txt = ax.text(r_inside * x, r_inside * y, bold(tex(label)) + '\n' + bold(tail), ha='center',
                      va='center', fontsize=9, color=on_wedge_ink(color))
        corners = ax.transData.inverted().transform(txt.get_window_extent(renderer).corners())
        if all(in_wedge(c, wedge) for c in corners):
            continue
        txt.remove()
        branches.append((x, y, bold(tex(label)) + '\n' + bold(tail) if branch_sep == '\n'
                         else bold(tex(label) + branch_sep + tail)))

    # Sit each branch label off the rim beside its own wedge, dropping it only as far as it
    # takes to clear the label above, so neither labels nor leaders can collide.
    step = min(0.30 if branch_sep == '\n' else 0.18, 1.9 * R_LABEL / max(len(branches), 1))
    for side in (1, -1):
        y_text = R_LABEL * radius - 0.04 if stack is None else stack.get(side, R_LABEL - 0.04)
        for x, y, text in sorted((b for b in branches if (1 if b[0] >= 0 else -1) == side),
                                 key=lambda b: -b[1]):
            y_text = min(y_text, R_LABEL * radius * y)
            # Hold the label over its own wedge, but never let the anchor fall inside the rim.
            x_text = (np.sign(x) or 1) * max(abs(R_LABEL * radius * x),
                                             np.sqrt(max((1.3 * radius) ** 2 - y_text ** 2, 0)))
            # The text reads away from its anchor; if that would take it back across the pie,
            # lift it clear of the rim so it passes over the top instead.
            if np.sign(x_text) == -label_side and abs(y_text) < 1.06 * radius:
                y_text = (1.06 if y_text >= 0 else -1.06) * radius
            # Short radial stub off the wedge, one break, then straight to the label. Pushing
            # outwards only helps when the label is further out; for one sitting below its own
            # slice a long stub just spikes up before coming back down across its neighbours.
            stub = STUB if y_text >= radius * y else 1.05
            ax.plot([radius * x, stub * radius * x, x_text], [radius * y, stub * radius * y, y_text],
                    color=LEADER, lw=0.7, clip_on=False, solid_capstyle='round')
            ax.text(x_text + label_side * 0.04, y_text, text,
                    ha='left' if label_side > 0 else 'right', va='center', fontsize=9, color=INK)
            y_text -= step
        if stack is not None:
            stack[side] = min(y_text, stack.get(side, y_text))
    return wedges


def link_slice(ax_from, wedge, ax_to):
    """Two tangent lines opening one wedge out to span the whole pie on ax_to."""
    for theta, sign in ((wedge.theta2, -1), (wedge.theta1, 1)):
        rim = np.array([np.cos(np.deg2rad(theta)), np.sin(np.deg2rad(theta))])
        start = rim  # break clear of its own rim first, like the label leaders
        ax_from.plot(*zip(rim, start), color=LEADER, lw=0.7, zorder=0, clip_on=False)
        # Where that point falls in the other pie's coordinates.
        sx, sy = ax_to.transData.inverted().transform(ax_from.transData.transform(start))
        d = np.hypot(sx, sy)
        if d <= 1:
            continue
        phi = np.arctan2(sy, sx) + sign * np.arccos(1 / d)  # tangent point on the unit rim
        ax_to.add_artist(ConnectionPatch(
            xyA=start, coordsA=ax_from.transData,
            xyB=(np.cos(phi), np.sin(phi)), coordsB=ax_to.transData,
            color=LEADER, lw=1.5, zorder=0, clip_on=False,
        ))


desig = as_str(in_both_srcs.data['tns_designation'])
types = as_str(in_both_srcs.data['tns_type'])
named = desig != ''

# tns_designation is a unique name per object, so pie its presence, then type up the named ones.
match = Counter({'TNS matched': int(named.sum()), 'No TNS match': int((~named).sum())})

by_type = Counter(dict(Counter(t for t in types[named] if t).most_common()))
if (n_unclassified := int(named.sum()) - sum(by_type.values())):
    by_type['Unclassified'] = n_unclassified

fig, axes = plt.subplots(1, 2, figsize=(10, 5), facecolor='white')
# Left pie starts 45 deg round so the matched sliver faces the right pie, not straight up.
match_wedges = pie(axes[0], match, [PALETTE[0], GRAY], startangle=80, label_side=-1)
pie(axes[1], by_type, PALETTE[:len(by_type) - 1] + [GRAY])
fig.subplots_adjust(wspace=0.2)
link_slice(axes[0], match_wedges[0], axes[1])  # needs the final axes positions
plt.show()
```

## WISE Colors

```python
# Per-requirement masks for the Stern et al. (2012, ApJ 753, 30) WISE AGN criterion.
# Nothing is applied here: every condition comes back separately so it can be
# inspected, combined, or ignored. Stern's own requirements are kept distinct
# from additions that are not his.
#
#   W1 - W2 >= 0.8 (Vega)    red mid-IR colour from AGN-heated dust; stars and
#                            ordinary galaxies sit near 0 on the Rayleigh-Jeans tail
#   SNR >= 10 in W2          the same cut as W2 <= 15.05, just in other units
#   blend flag nb <= 2       "relatively isolated sources"
#   no D/P/H/O in W1 or W2   diffraction spike / persistence / halo / optical ghost
#
# Reuses as_str() from the TNS section above.


def as_float(col):
    """Column -> float array with NaN where missing, masked or not."""
    return np.asarray(col.filled(np.nan) if hasattr(col, 'filled') else col, dtype=float)


def stern12_masks(data, snr_min=10.0, colour_min=0.8, nb_max=2, n_sigma=2.0):
    """Named boolean masks, one per condition, over `Sources.data`.

    Keys prefixed `stern_` are the paper's requirements; `extra_` are not. Each
    mask tests exactly one thing and is *not* gated on having a WISE match, so
    unmatched sources pass the `extra_` ones trivially -- combine with
    `stern_has_match` before reading any count as physical.
    """
    w1, w2 = as_float(data['wise_w1mpro']), as_float(data['wise_w2mpro'])
    e1, e2 = as_float(data['wise_w1sigmpro']), as_float(data['wise_w2sigmpro'])
    colour, colour_err = w1 - w2, np.hypot(e1, e2)
    cc = as_str(data['wise_cc_flags'])

    return {
        # --- Stern et al. 2012 ------------------------------------------------
        'stern_has_match':      np.asarray(data['wise_matched'], dtype=bool),
        'stern_red':            colour >= colour_min,
        'stern_w2_snr':         as_float(data['wise_w2snr']) >= snr_min,
        'stern_unblended':      as_float(data['wise_nb']) <= nb_max,
        # cc_flags is one character per band in W1 W2 W3 W4 order; only the first
        # two matter, so a source flagged solely at 12 or 22 um is kept.
        'stern_uncontaminated': np.array([len(c) >= 2 and c[:2] == '00' for c in cc]),

        # --- not Stern's, offered separately ----------------------------------
        # The paper notes most sources are also >=10 sigma in W1, but does not
        # require it, so it lives here rather than above.
        'extra_w1_snr':         as_float(data['wise_w1snr']) >= snr_min,
        # ph_qual 'U' means SNR < 2, i.e. the magnitude is a 95% upper limit. A
        # W2 limit makes the quoted colour an upper bound, so it can only push a
        # source spuriously *over* 0.8, never under.
        'extra_w2_detected':    as_str(data['wise_w2_ph_qual']) != 'U',
        # Require the colour to clear the threshold significantly: a source at
        # 0.82 +/- 0.09 is not a detection of red colour.
        'extra_red_nsigma':     (colour - n_sigma * colour_err) >= colour_min,
        # Nothing else inside the ~6" W1 beam (our own neighbour count within
        # 8", not a WISE flag) -- blending that nb alone does not always catch.
        'extra_isolated_beam':  as_float(data['wise_n_in_beam']) <= 1,
        # Resolved sources: profile-fit magnitudes underestimate their flux.
        'extra_not_extended':   as_float(data['wise_ext_flg']) == 0,
    }

in_both_masks = stern12_masks(in_both_srcs.data)
in_ztf_masks = stern12_masks(in_ztf_srcs.data)
```

```python
# Where the in-both sources fall against the Stern+12 WISE cut. The four buckets are
# mutually exclusive and sum to the sample: a source has no WISE match, or has one that
# fails the paper's quality requirements, or clears them and is red, or clears them and is not.
matched = in_both_masks['stern_has_match']
quality = np.logical_and.reduce([m for k, m in in_both_masks.items()
                                 if k.startswith('stern_')
                                 and k not in ('stern_has_match', 'stern_red')])
red = in_both_masks['stern_red']

wise = Counter({
    fr'$\rm{{W}}1 - \rm{{W}}2 \geq 0.8$': int((matched & quality & red).sum()),
    fr'$\rm{{W}}1 - \rm{{W}}2 < 0.8$': int((matched & quality & ~red).sum()),
    'Fails Stern+12 cuts': int((matched & ~quality).sum()),
    'No WISE match': int((~matched).sum()),
})
wise = Counter({k: v for k, v in wise.items() if v})  # a pie has no use for empty wedges

fig, ax = plt.subplots(figsize=(5.5, 5), facecolor='white')
pie(ax, wise, [PALETTE[1], PALETTE[0], PALETTE[2], GRAY], startangle=90, branch_sep='\n')
plt.show()

# Plot the WISE Color distributions
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.hist(
    in_both_srcs.data['wise_w1mpro'][matched & quality] - in_both_srcs.data['wise_w2mpro'][matched & quality],
    bins=50,
    color='k',
    label='Stern+12 Quality'
)
ax.hist(
    in_both_srcs.data['wise_w1mpro'][matched] - in_both_srcs.data['wise_w2mpro'][matched],
    bins=50,
    color='gray',
    zorder=-1,
    label='All Matches'
)
ax.axvline(0.8, color='red', label=fr'$\rm{{W}}1 - \rm{{W2}} = 0.8$')

ax.set_xlabel(bold(fr'$\rm{{W}}1 - \rm{{W2}}$'), fontsize=14)
ax.set_ylabel(bold('Count'), fontsize=14)
ax.legend()

plt.show()
```

## AGN-DB Breakdown

```python
import json

# AGN-DB carries one class per contributing catalog. Collapse each source's list to a single
# label: every AGN subtype counts as AGN, catalogs that agree give their class, catalogs that
# disagree give 'Multiple Classes'. 'unknown' is the absence of a class rather than one of
# them, so it never counts as disagreement -- only a source no catalog classifies is
# Unclassified.
AGN_TYPES = {'generic_agn', 'type1', 'type2', 'blazar'}
CLASS_LABELS = {'star': 'Star', 'galaxy': 'Galaxy', 'lens_system': 'Lens system'}


def agn_db_class(entry):
    """One display class for a source, or '' where AGN-DB only ever says 'unknown'."""
    claims = set(json.loads(entry)) if entry.startswith('[') else ({entry} if entry else set())
    labels = {'AGN' if c in AGN_TYPES else CLASS_LABELS.get(c, c.title())
              for c in claims - {'unknown'}}
    if not labels:
        return ''
    return labels.pop() if len(labels) == 1 else 'Multiple Classes'


matched = np.asarray(in_both_srcs.data['agn_db_matched'], dtype=bool)
classes = np.array([agn_db_class(e) for e in as_str(in_both_srcs.data['agn_db_best_class_all'])])

match = Counter({'AGN-DB matched': int(matched.sum()), 'No AGN-DB match': int((~matched).sum())})

by_class = Counter(dict(Counter(c for c in classes[matched] if c).most_common()))
if (n_unclassified := int(matched.sum()) - sum(by_class.values())):
    by_class['Unclassified'] = n_unclassified
colors = PALETTE[:len(by_class)]
if 'Unclassified' in by_class:
    colors[-1] = GRAY

fig, axes = plt.subplots(1, 2, figsize=(10, 5), facecolor='white')
# Left pie starts 80 deg round so the matched slice faces the right pie, not straight up.
match_wedges = pie(axes[0], match, [PALETTE[0], GRAY], startangle=80, label_side=-1)
pie(axes[1], by_class, colors)
fig.subplots_adjust(wspace=0.2)
link_slice(axes[0], match_wedges[0], axes[1])  # needs the final axes positions
plt.show()
```

## Unified Classification

```python
# Every catalog's verdict on every candidate, pooled into one group per source. A WISE match
# alone does not count as being seen: WISE gives colours, not classes, so only its Stern+12
# verdict matters. With no votes a source is 'Unclassified' if some catalog saw it and
# 'Unmatched' if none did. Otherwise the votes are reconciled by `unified_group`, which is
# where catalogs describing the same object in different words stop counting as disagreement.
UNIFIED_SPEC = {'QSO': 'AGN', 'GALAXY': 'Galaxy', 'STAR': 'Star'}
SPEC_CATALOGS = ('SDSS', 'DESI')
TNS_AGN_TYPES = {'AGN', 'QSO'}   # TNS types that name the host, not a transient
MATCH_COLS = ('tns_matched', 'agn_db_matched', 'sdss_matched', 'desi_matched', 'simbad_matched')
DIM = '#a8a79e'  # second neutral, so the two no-information groups stay distinguishable
# Fixed group -> colour and wedge order, so the two panels stay comparable: a group keeps its
# colour and its place in the ring even where the two samples rank it differently.
GROUP_COLORS = {'AGN': PALETTE[0], 'Galaxy': PALETTE[1], 'Transient': PALETTE[2],
                'Star': PALETTE[3], 'Multiple Classes': PALETTE[4],
                'Unclassified': GRAY, 'Unmatched': DIM}


def unified_votes(data):
    """Per source, {catalog: set of groups it voted for}; empty where no catalog said anything."""
    votes = [{} for _ in range(len(data))]

    def vote(i, catalog, group):
        votes[i].setdefault(catalog, set()).add(group)

    # TNS: a designation means somebody reported a transient here -- unless TNS typed it as
    # an AGN or QSO, which names what is there rather than something that happened.
    tns_type = as_str(data['tns_type'])
    for i in np.flatnonzero(as_str(data['tns_designation']) != ''):
        vote(i, 'TNS', 'AGN' if tns_type[i] in TNS_AGN_TYPES else 'Transient')

    # WISE: only clearing Stern+12 outright is a statement. A blue mid-IR colour says 'not an
    # AGN by this test', which is not a classification, so it votes for nothing.
    masks = stern12_masks(data)
    for i in np.flatnonzero(np.logical_and.reduce([m for k, m in masks.items()
                                                   if k.startswith('stern_')])):
        vote(i, 'WISE', 'AGN')

    # AGN-DB: each contributing catalog votes separately, so a source can disagree with itself.
    # 'unknown' and 'lens_system' state no class, so neither votes.
    agn_cls = as_str(data['agn_db_best_class_all'])
    for i in np.flatnonzero(np.asarray(data['agn_db_matched'], dtype=bool)):
        s = agn_cls[i]
        for c in set(json.loads(s)) if s.startswith('[') else ({s} if s else set()):
            if c in AGN_TYPES:
                vote(i, 'AGN-DB', 'AGN')
            elif c in ('star', 'galaxy'):
                vote(i, 'AGN-DB', CLASS_LABELS[c])

    # SDSS and DESI spectroscopic classes.
    for catalog, matched_col, class_col in (('SDSS', 'sdss_matched', 'sdss_class'),
                                            ('DESI', 'desi_matched', 'desi_spectype')):
        spec = as_str(data[class_col])
        for i in np.flatnonzero(np.asarray(data[matched_col], dtype=bool)):
            if spec[i] in UNIFIED_SPEC:
                vote(i, catalog, UNIFIED_SPEC[spec[i]])

    # SIMBAD: its object-type hierarchy picks the group. Candidates ('AG?', 'Q?', ...) do
    # not vote, and neither do types that say nothing about the class (radio, X-ray, 'blue
    # object', HII region, cluster, unknown).
    path = as_str(data['simbad_otype_path'])
    candidate = np.asarray(data['simbad_is_candidate'].filled(False)
                           if hasattr(data['simbad_is_candidate'], 'filled')
                           else data['simbad_is_candidate'], dtype=bool)
    for i in np.flatnonzero(np.asarray(data['simbad_matched'], dtype=bool) & ~candidate):
        if group := simbad_group(path[i]):
            vote(i, 'SIMBAD', group)
    return votes


def simbad_group(path):
    """The group a SIMBAD otype path ('G > AGN > QSO', '* > SN*', ...) belongs to, or ''."""
    levels = [p.strip() for p in path.split('>')]
    if 'AGN' in levels:
        return 'AGN'
    if levels[0] == 'G':
        return 'Galaxy'
    if 'SN*' in levels or levels[0] == 'ev':
        return 'Transient'
    if levels[0] == '*':
        return 'Star'
    return ''


def unified_group(votes):
    """One group from one source's {catalog: groups}, or '' if nobody voted.

    Rules, in order:
      Transient wins. A transient at an AGN or in a galaxy is the interesting part.
      Spectra outrank a generic 'star'. SIMBAD's plain 'Star' is often just an unresolved
        point source and AGN-DB's 'star' a flag, so either yields to an SDSS/DESI spectrum
        that says something else.
      AGN beats Galaxy. SDSS/DESI call any AGN without broad lines GALAXY, so 'Galaxy' plus
        an AGN claim is an AGN in its host -- unless the spectra themselves disagree (one
        says QSO, one says GALAXY), which is a real conflict.
    """
    groups = set().union(*votes.values()) if votes else set()
    spec = set().union(*(votes.get(c, set()) for c in SPEC_CATALOGS))
    if 'Transient' in groups:
        return 'Transient'
    if spec and 'Star' not in spec:
        groups.discard('Star')
    if {'AGN', 'Galaxy'} <= groups and not {'AGN', 'Galaxy'} <= spec:
        groups.discard('Galaxy')
    if not groups:
        return ''
    return next(iter(groups)) if len(groups) == 1 else 'Multiple Classes'


def unified_groups(data):
    """One group per source, counted up in the fixed GROUP_COLORS order."""
    seen = np.logical_or.reduce([np.asarray(data[c], dtype=bool) for c in MATCH_COLS])
    tally = Counter()
    for v, in_a_catalog in zip(unified_votes(data), seen):
        tally[unified_group(v) or ('Unclassified' if in_a_catalog else 'Unmatched')] += 1
    return Counter({g: tally[g] for g in GROUP_COLORS if tally[g]})


fig, axes = plt.subplots(1, 2, figsize=(10, 5), facecolor='white')
for ax, srcs, name, side, start_angle in ((axes[0], in_both_srcs, 'In both', -1, 30),
                             (axes[1], in_ztf_srcs, 'In ZTF only', 1, 40)):
    groups = unified_groups(srcs.data)
    pie(ax, groups, [GROUP_COLORS[g] for g in groups], label_side=side, branch_sep='\n', startangle=start_angle)
    ax.set_title(bold(f'{name} (n={len(srcs.data):,})'), fontsize=11, color=INK, y=0.9)
fig.subplots_adjust(wspace=0.1)
plt.savefig('/Users/adamboesky/Research/long_transients/Figures/catalog_matched_demographics_pie_charts.pdf', bbox_inches='tight')
plt.show()
```

```python
# The unified pie exactly as above, plus one extra ring sitting outside a single slice --
# Unclassified -- splitting it by what WISE had to say about those sources. Every other group
# is untouched: the ring simply has nothing to add about them.
UNCLASSIFIED_SPLIT = ('No WISE match', 'Fails quality cuts',
                      r'$\rm{W}1 - \rm{W}2 \geq 0.8$', r'$\rm{W}1 - \rm{W}2 < 0.8$')
# Shades of the parent's neutral: subdividing one group is a sequential scale, not a new set
# of categories.
SPLIT_SHADES = ['#e2e1da', '#c8c7bd', '#aeada3', '#949389']
RING = 0.22  # thickness of the ring riding outside the Unclassified slice


def unclassified_wise_split(data):
    """How the Unclassified sources of `data` break down by what WISE could say about them."""
    seen = np.logical_or.reduce([np.asarray(data[c], dtype=bool) for c in MATCH_COLS])
    unc = np.array([not v and s for v, s in zip(unified_votes(data), seen)])

    masks = stern12_masks(data)
    matched, red = masks['stern_has_match'], masks['stern_red']
    quality = np.logical_and.reduce([m for k, m in masks.items() if k.startswith('stern_')
                                     and k not in ('stern_has_match', 'stern_red')])
    parts = (~matched, matched & ~quality, matched & quality & red, matched & quality & ~red)
    split = Counter({name: int((unc & part).sum())
                     for name, part in zip(UNCLASSIFIED_SPLIT, parts)})
    return Counter({k: v for k, v in split.items() if v})  # a pie has no use for empty wedges


def mid_y(counts, startangle, key):
    """Height of a wedge's midpoint, for deciding which labels get the top of the queue."""
    before = sum(v for k, v in counts.items() if list(counts).index(k) < list(counts).index(key))
    mid = startangle - 360 * (before + counts[key] / 2) / sum(counts.values())
    return np.sin(np.deg2rad(mid))


# Each panel is turned so every wedge that needs an outside label sits in the top corner its
# labels read towards -- otherwise they all queue below a wedge at y~0 and pile up in a corner.
fig, axes = plt.subplots(1, 2, figsize=(12, 6), facecolor='white')
for ax, srcs, name, side, start_angle in ((axes[0], in_both_srcs, 'In both', -1, 20),
                                          (axes[1], in_ztf_srcs, 'In ZTF only', 1, 0)):
    groups = unified_groups(srcs.data)
    split = unclassified_wise_split(srcs.data)

    # The ring covers the whole circle so it stays aligned with the pie, but everything either
    # side of Unclassified is a blank, colourless spacer: drawn as nothing, labelled as nothing.
    order = list(groups)
    cut = order.index('Unclassified')
    ring = Counter({' ': sum(groups[g] for g in order[:cut])})
    ring.update(split)
    ring['  '] = sum(groups[g] for g in order[cut + 1:])
    ring_colors = ['none'] + SPLIT_SHADES[:len(split)] + ['none']

    # One shared label queue so the pie's labels and the ring's never land on each other. The
    # queue fills from the top down, so whichever ring holds the higher slices goes first.
    stack = {}
    draw_ring = lambda: pie(ax, ring, ring_colors, label_side=side, branch_sep='--',
                            startangle=start_angle, width=RING, radius=1 + RING, stack=stack)
    draw_pie = lambda: pie(ax, groups, [GROUP_COLORS[g] for g in groups], label_side=side,
                           branch_sep='--', startangle=start_angle, stack=stack)
    # Compare where each leader *starts*, not where its slice is: the ring sits further out, so
    # its leaders begin higher. Getting this backwards sends one raking down across the labels.
    if len(split) < 2:  # one part is no subdivision at all; the ring would just trace the slice
        pie(ax, groups, [GROUP_COLORS[g] for g in groups], label_side=side, branch_sep='\n',
            startangle=start_angle)
        ax.set_title(bold(f'{name} (n={len(srcs.data):,})'), fontsize=11, color=INK, pad=14)
        continue

    highest_slice = max((mid_y(groups, start_angle, g) for g, n in groups.items()
                         if 100 * n / sum(groups.values()) <= SMALL), default=-1)
    ring_start = (1 + RING) * mid_y(groups, start_angle, 'Unclassified')
    order = (draw_ring, draw_pie) if ring_start > highest_slice else (draw_pie, draw_ring)
    for draw in order:
        draw()
    ax.set_xlim(*(np.array(ax.get_xlim()) * (1 + RING)))  # room for the ring and its labels
    ax.set_ylim(*(np.array(ax.get_ylim()) * (1 + RING)))
    ax.set_title(bold(f'{name} (n={len(srcs.data):,})'), fontsize=11, color=INK, pad=14)
fig.subplots_adjust(wspace=0.2)
plt.show()
```

## SPHEREx vs WISE photometry

```python
# SPHEREx vs WISE photometry for a handful of AGN-DB AGN.
#
# SPHEREx has no source catalogue: IRSA serves only Level-2 spectral images, where a source's
# wavelength depends on where it lands on the linear variable filter. So a "query" here means
# forced aperture photometry -- find every image covering the position, read the wavelength for
# that pixel out of the image's own WCS-WAVE lookup table, and measure a flux. One image gives
# one spectral point; sampling several per detector walks across the 0.75-5 um range.
#
# The fluxes are aperture photometry with no PSF or aperture correction, so treat them as good
# to tens of percent, not calibrated. The WISE points in the overlap region are the check.
import concurrent.futures
from astropy.coordinates import SkyCoord
from astropy.io import fits
from astropy.table import Table
from astropy.utils.data import conf
from astropy.wcs import WCS
import astropy.units as u
from astroquery.ipac.irsa import Irsa
from photutils.aperture import CircularAperture, CircularAnnulus, aperture_photometry
from scipy.interpolate import RegularGridInterpolator

conf.remote_timeout = 120  # SPHEREx MEFs are large and read lazily over HTTP

SA_ARCSEC2 = 6.15 ** 2                                        # IRSA's default pixel solid angle
MJY_SR_TO_MJY_PIX = SA_ARCSEC2 * (np.pi / (180 * 3600)) ** 2 * 1e9   # MJy/sr -> mJy/pixel
AP_R, ANN_IN, ANN_OUT = 3.0, 5.0, 9.0                         # pixels; SPHEREx pixels are 6.15"
STAMP = 12                                                    # half-size of the cutout we read

# AllWISE: effective wavelength (um) and Vega zero point (Jy).
WISE_BANDS = {'W1': (3.3526, 309.540), 'W2': (4.6028, 171.787),
              'W3': (11.5608, 31.674), 'W4': (22.0883, 8.363)}


def _one_image(url, coord):
    """(wavelength um, bandwidth um, flux mJy) for one spectral image, or None."""
    try:
        with fits.open(url, use_fsspec=True) as hdul:
            header = hdul[1].header
            x, y = (float(v) for v in WCS(header).world_to_pixel(coord))
            if not (STAMP < x < header['NAXIS1'] - STAMP and STAMP < y < header['NAXIS2'] - STAMP):
                return None  # the source missed this detector on this pointing
            xi, yi = int(round(x)), int(round(y))
            stamp = hdul[1].section[yi - STAMP:yi + STAMP + 1,
                                    xi - STAMP:xi + STAMP + 1].astype(float)
            wave = Table(hdul['WCS-WAVE'].data)
    except Exception:
        return None  # a dropped connection is not worth failing the whole spectrum over

    # WCS-WAVE holds a coarse grid of control points; interpolate rather than snap to the
    # nearest, which on a ~200 pixel grid spacing would cost a few percent in wavelength.
    gx = np.asarray(wave['X'][0], float).ravel()
    gy = np.asarray(wave['Y'][0], float).ravel()
    values = np.asarray(wave['VALUES'][0], float)
    lam, band = (RegularGridInterpolator((gy, gx), values[..., k], bounds_error=False,
                                         fill_value=None)([[y, x]])[0] for k in (0, 1))

    centre = [(x - (xi - STAMP), y - (yi - STAMP))]
    annulus = CircularAnnulus(centre, r_in=ANN_IN, r_out=ANN_OUT).to_mask(method='center')[0]
    ring = annulus.multiply(stamp)
    sky = np.nanmedian(ring[(ring != 0) & np.isfinite(ring)])
    flux = aperture_photometry(stamp - sky, CircularAperture(centre, r=AP_R))['aperture_sum'][0]
    return float(lam), float(band), float(flux) * MJY_SR_TO_MJY_PIX


def spherex_spectrum(ra, dec, per_detector=5, workers=8):
    """Forced photometry spectrum at one position, sampling images across all six detectors."""
    coord = SkyCoord(ra, dec, unit='deg')
    hits = Irsa.query_sia(pos=(coord, 1 * u.arcsec), collection='spherex_qr2').to_table()
    hits = hits[hits['dataproduct_type'] == 'image']
    detector = np.array([str(b) for b in hits['energy_bandpassname']])
    urls = [str(hits['access_url'][i]) for d in sorted(set(detector))
            for i in np.flatnonzero(detector == d)[:per_detector]]

    with concurrent.futures.ThreadPoolExecutor(workers) as pool:  # network-bound, so thread it
        rows = [r for r in pool.map(lambda u_: _one_image(u_, coord), urls) if r]
    rows = [r for r in rows if np.isfinite(r[2])]  # masked or off-edge pixels give NaN
    out = Table(rows=rows, names=('wavelength', 'bandwidth', 'flux')) if rows else \
        Table(names=('wavelength', 'bandwidth', 'flux'))
    out.sort('wavelength')
    return out


def wise_photometry(row):
    """WISE W1-W4 as (wavelength um, flux mJy, error mJy), skipping bands with no measurement."""
    out = []
    for band, (lam, zero_point) in WISE_BANDS.items():
        mag, err = row[f'wise_{band.lower()}mpro'], row[f'wise_{band.lower()}sigmpro']
        if mag is np.ma.masked or not np.isfinite(float(mag)):
            continue
        flux = zero_point * 10 ** (-float(mag) / 2.5) * 1e3  # Jy -> mJy
        err = flux * 0.4 * np.log(10) * float(err) if np.isfinite(float(err)) else np.nan
        out.append((lam, flux, err))
    return np.array(out).T if out else np.empty((3, 0))


# --- pick the targets and draw -----------------------------------------------------------
N_AGN, PER_DETECTOR = 3, 4

data = in_both_srcs.data
is_agn = np.array([agn_db_class(e) == 'AGN'
                   for e in as_str(data['agn_db_best_class_all'])])
usable = (is_agn & np.asarray(data['agn_db_matched'], dtype=bool)
          & np.isfinite(as_float(data['wise_w3mpro'])) & (as_float(data['wise_w2snr']) >= 20))
targets = np.flatnonzero(usable)[np.argsort(as_float(data['wise_w1mpro'])[usable])][:N_AGN]

fig, axes = plt.subplots(1, N_AGN, figsize=(4.2 * N_AGN, 4.2), facecolor='white', sharey=False)
for ax, i in zip(np.atleast_1d(axes), targets):
    row = data[int(i)]
    sx = spherex_spectrum(float(row['ra']), float(row['dec']), per_detector=PER_DETECTOR)
    lam, flux, err = wise_photometry(row)

    ax.errorbar(sx['wavelength'], sx['flux'], xerr=sx['bandwidth'] / 2, fmt='o', ms=3.5,
                color=PALETTE[0], lw=0.8, label=bold('SPHEREx (forced)'), zorder=2)
    ax.errorbar(lam, flux, yerr=err, fmt='s', ms=7, color=PALETTE[1], lw=1.2,
                label=bold('AllWISE'), zorder=3)

    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_xlabel(bold(r'$\lambda$ [$\mu$m]'), fontsize=12)
    # WISE profile-fit magnitudes assume a point source; where the host is resolved at low z our
    # aperture sums in more light than they do, so expect an offset that is not a calibration error.
    z = float(row['agn_db_best_z_merged'])
    ax.set_title(bold(tex(f'{float(row["ra"]):.4f}{float(row["dec"]):+.4f}')
                      + (rf'  $z={z:.3f}$' if np.isfinite(z) else '')),
                 fontsize=10, color=INK)
np.atleast_1d(axes)[0].set_ylabel(bold(r'$F_\nu$ [mJy]'), fontsize=12)
np.atleast_1d(axes)[0].legend(frameon=False, fontsize=9, loc='upper left')
fig.tight_layout()
plt.show()
```
